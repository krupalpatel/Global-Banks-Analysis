```python
# =====================================================
# 1. IMPORT LIBRARIES
# =====================================================

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

plt.style.use('default')
sns.set_style("whitegrid")

import plotly.io as pio

pio.renderers.default = 'iframe'
```

```python
# =====================================================
# 2. LOAD DATA
# =====================================================
file_path = "/kaggle/input/datasets/krupalpatel07/bank-of-america-daily-historical-data/bofa.csv"
df = pd.read_csv(file_path)
```

```python
# =====================================================
# 3. PREPROCESSING
# =====================================================
df.columns = [col.lower() for col in df.columns]
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')
df.set_index('date', inplace=True)
```

```python
# =====================================================
# 4. HTML
# =====================================================
from IPython.display import display, HTML

def luxury_header(text):
    display(HTML(f"""
    <div style="
        background: linear-gradient(90deg, #fdfcfb, #e2d1c3);
        padding: 20px;
        border-radius: 12px;
        border: 1px solid #d4af37;
        margin-top:20px;
    ">
        <h1 style="color:#b8962e; text-align:center; font-family:Georgia;">
            {text}
        </h1>
    </div>
    """))

luxury_header("📊 Price Structure Overview")
```

```python
luxury_header("🔥 Market Pulse (Trend + Momentum)")

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['close'],
    mode='lines',
    name='Close Price',
    line=dict(width=2),
))

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['close'].rolling(50).mean(),
    mode='lines',
    name='50D MA',
    line=dict(dash='dash')
))

fig.update_layout(
    template='plotly_dark',
    title="Bank of America Trend Flow",
    xaxis_title="Date",
    yaxis_title="Price",
    hovermode='x unified'
)

fig.show()
```

```python
# =====================================================
# 5. CLEAN PRICE VISUAL
# =====================================================
plt.figure(figsize=(14,6))
plt.plot(df['close'])
plt.title("Closing Price Trend", fontsize=16)
plt.show()
```

```python
# =====================================================
# 6. MARKET MICROSTRUCTURE ANALYSIS
# =====================================================
luxury_header("📉 Intraday Spread & Efficiency")

# High-Low Spread
df['spread'] = df['high'] - df['low']

# Efficiency Ratio
df['efficiency'] = abs(df['close'] - df['open']) / (df['high'] - df['low'])

fig = px.line(df, y='spread', title="Daily Spread")
fig.show()
```

```python
# =====================================================
# 7. VOLUME PRESSURE ANALYSIS
# =====================================================
luxury_header("📦 Volume Pressure Zones")

# Volume Weighted Price Proxy
df['vwap_proxy'] = (df['close'] * df['volume']).cumsum() / df['volume'].cumsum()

fig = go.Figure()
fig.add_trace(go.Scatter(x=df.index, y=df['close'], name='Close'))
fig.add_trace(go.Scatter(x=df.index, y=df['vwap_proxy'], name='VWAP Proxy'))
fig.show()
```

```python
# =====================================================
# 8. ALPHA FACTOR ENGINEERING
# =====================================================
luxury_header("🧠 Alpha Signal Engineering")

# Momentum
df['momentum'] = df['close'].pct_change(10)

# Mean Reversion
df['zscore'] = (df['close'] - df['close'].rolling(20).mean()) / df['close'].rolling(20).std()

fig = px.scatter(df, x='momentum', y='zscore')
fig.show()
```

```python
# =====================================================
# 9. LINEAR MODEL (ML)
# =====================================================
luxury_header("📐 Linear Alpha Model")

# Prepare data
df = df.dropna()
X = df[['momentum','zscore']]
y = df['close']

# Split
split = int(len(df)*0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# Model
model = LinearRegression()
model.fit(X_train, y_train)

preds = model.predict(X_test)

print("R2 Score:", r2_score(y_test, preds))

# Plot
plt.figure(figsize=(12,5))
plt.plot(y_test.values, label='Actual')
plt.plot(preds, label='Predicted')
plt.legend()
plt.title("Model Fit")
plt.show()
```

```python
# =====================================================
# 10. DISTRIBUTION ANALYSIS
# =====================================================
luxury_header("📊 Return Distribution (Risk View)")

df['returns'] = df['close'].pct_change()

plt.figure(figsize=(10,5))
sns.histplot(df['returns'], bins=50, kde=True)
plt.title("Returns Distribution")
plt.show()
```

```python
luxury_header("🧠 Linear Regression Insight")

df_lr = df.copy()
df_lr['time_index'] = range(len(df_lr))

X = df_lr[['time_index']]
y = df_lr['close']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)

fig = go.Figure()

fig.add_trace(go.Scatter(x=df_lr.index, y=y, name="Actual"))
fig.add_trace(go.Scatter(x=df_lr.index, y=y_pred, name="Trend Line"))

fig.update_layout(
    template='plotly_white',
    title=f"Regression Fit (R² = {r2:.4f})"
)

fig.show()
```

```python
print(f"📌 Model Strength (R²): {r2:.4f}")
```

```python
luxury_header("🌊 Volatility Heatwave")

df['returns'] = df['close'].pct_change()
df['volatility'] = df['returns'].rolling(21).std()

fig = px.imshow(
    [df['volatility'].fillna(0).values],
    aspect='auto',
    color_continuous_scale='viridis'
)

fig.update_layout(
    title="Volatility Intensity Map (21D Rolling)",
    yaxis_showticklabels=False
)

fig.show()
```

```python
luxury_header("📊 Candlestick + Volume")

fig = go.Figure()

fig.add_trace(go.Candlestick(
    x=df.index,
    open=df['open'],
    high=df['high'],
    low=df['low'],
    close=df['close'],
    name='Price'
))

fig.add_trace(go.Bar(
    x=df.index,
    y=df['volume'],
    name='Volume',
    yaxis='y2',
    opacity=0.3
))

fig.update_layout(
    template='plotly_dark',
    yaxis2=dict(
        overlaying='y',
        side='right',
        showgrid=False
    ),
    title="Price Action with Volume Pressure"
)

fig.show()
```

```python
luxury_header("🧬 Alpha Signals (Feature Engineering)")

df['ma_20'] = df['close'].rolling(20).mean()
df['ma_100'] = df['close'].rolling(100).mean()

df['momentum'] = df['close'] - df['close'].shift(10)

fig = go.Figure()

fig.add_trace(go.Scatter(x=df.index, y=df['ma_20'], name="MA 20"))
fig.add_trace(go.Scatter(x=df.index, y=df['ma_100'], name="MA 100"))

fig.update_layout(
    title="Moving Average Regime Detection",
    template='plotly_white'
)

fig.show()
```
